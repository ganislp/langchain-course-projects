```python
code_prompt = PromptTemplate(
    input_variables=["language","task"],
    template="Write a very shot {language} function that will {task}"
)
test_prompt = PromptTemplate(
    input_variables=['language','code'],template="Write a test for the following {language} /n code:{code}"
)

llm = ChatOpenAI(model="gpt-5.4-mini")
code_chain = code_prompt | llm | parser
test_chain = test_prompt | llm | parser

overall  =(
    RunnablePassthrough.assign(code=code_chain)|
    RunnablePassthrough.assign(test=test_chain)
)
result = overall.invoke({
    "language": "python",
    "task": "return a list of numbers",
})

print("=== CODE ===")
print(result["code"])
print("\n=== TEST ===")
print(result["test"])
```
