```python
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnablePassthrough
```

```python
load_dotenv(override=True)
```

```python
parser = RunnablePassthrough()

recipe_prompt = PromptTemplate(
    input_variables=['dish'],
    template="Write a simple recipe for {dish} for {servings} people."
)

ingredient_prompt = PromptTemplate(
    input_variables=['recipe'],
    template="List only the ingredients with quantities from this recipe,one per line:\n{recipe}"
)

llm = ChatOpenAI(model="gpt-5.4-mini")

recipe_chain = recipe_prompt | llm | parser
ingredient_chain = ingredient_prompt | llm | parser

cooking_overall = (
    RunnablePassthrough.assign(recipe=recipe_prompt)|
    RunnablePassthrough.assign(ingredients=ingredient_chain)
)

result = cooking_overall.invoke({"dish": "Banoffee Pie", "servings": 6})

print("=== RECIPE ===")
print(result["recipe"])
print("\n=== INGREDIENTS ===")
print(result["ingredients"])

```
